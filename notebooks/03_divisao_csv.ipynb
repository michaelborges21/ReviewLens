```python
import pandas as pd

def dividir_csv(arquivo_entrada, num_partes):
  """
  Divide um arquivo CSV em várias partes.

  Args:
    arquivo_entrada: Caminho para o arquivo CSV de entrada.
    num_partes: Número de partes em que o arquivo deve ser dividido.
  """

  try:
    # Lê o arquivo CSV com pandas
    df = pd.read_csv(arquivo_entrada)

    # Calcula o número de linhas por parte
    linhas_por_parte = len(df) // num_partes

    # Divide o DataFrame em partes
    for i in range(num_partes):
      inicio = i * linhas_por_parte
      fim = (i + 1) * linhas_por_parte if i < num_partes - 1 else len(df)
      df_parte = df[inicio:fim]

      # Salva cada parte em um novo arquivo CSV
      nome_arquivo_saida = f"{arquivo_entrada[:-4]}_{i+1}.csv"
      df_parte.to_csv(nome_arquivo_saida, index=False)

    print(f"Arquivo '{arquivo_entrada}' dividido com sucesso em {num_partes} partes.")

  except FileNotFoundError:
    print(f"Erro: Arquivo '{arquivo_entrada}' não encontrado.")
  except Exception as e:
    print(f"Erro ao dividir o arquivo: {e}")


if __name__ == "__main__":
  arquivo_entrada = input("Digite o caminho para o arquivo CSV: ")
  num_partes = int(input("Digite o número de partes: "))
  dividir_csv(arquivo_entrada, num_partes)
```
